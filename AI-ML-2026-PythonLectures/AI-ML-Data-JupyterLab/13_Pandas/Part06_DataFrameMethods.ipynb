```python
import pandas as pd
```

```python
print("---Fetch data using csv files---")
df = pd.read_csv("Part5_employee_data.csv")
# print(df, type(df))
```

```python
#Methods

print("---head() -> it will print initial 5 values---")
print(df.head())

print("---tail() -> it will print last 2 values---")
print(df.tail(2))

print("---sample() -> it will fetch random sample---")
print(df.sample())

print("---sample() -> it will give you summary of data---")
print(df.info())

print("---shape -> it will give you shape of data ---")
print(df.shape)

print("---describe() -> it will give you summary/description min max mean and other data ---")
print(df.describe())

print("---nunique() -> it will give you count the number of unique (distinct) values in each column of a DataFrame. ---")
print(df.nunique())
```
