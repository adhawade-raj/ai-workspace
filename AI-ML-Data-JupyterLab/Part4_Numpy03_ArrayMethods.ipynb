```python
import numpy as np
```

```python
print("----# Setting up as int type---")
arr1 = np.zeros((2,3), dtype="int64")
print(arr1, arr1.shape)

arr2 = np.ones((5, ))
print(arr2, arr2.shape)

# prefill with val  
print("----# prefill with val---")
arr3 = np.full((3,4), 100)
print(arr3, arr3.shape) 

print("----# Identity matrix - digonal values same ---")
arr4 = np.eye((3))
print(arr4, arr4.shape)

print("----# Adding a range of values ---")
arr5 = np.arange(1,10)
print(arr5, arr5.shape)  

print("----# Adding a range of values ---")
arr6 = np.linspace(1,10, 4)
print(arr6, arr6.shape)  
```
