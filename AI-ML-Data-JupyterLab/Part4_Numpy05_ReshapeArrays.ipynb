```python
import numpy as np
```

```python
arr = np.array([[1,2,3],[4,5,6]])
print(arr, arr.shape)

print("--- After reshape ---")
reshaped = arr.reshape((3,2))
print(reshaped, reshaped.shape)
print()

print("----convert 2D array to 1D array---")
flattened = arr.flatten()
print(flattened, flattened.shape)
```
