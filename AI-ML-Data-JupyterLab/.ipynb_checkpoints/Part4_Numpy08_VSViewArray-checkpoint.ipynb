```python
import numpy as np
```

```python
nums = [1,2,3,4,5]
sub_list = nums[1:3]
print(sub_list)
sub_list[0] = 200
print(sub_list)
print(nums)

print("---Value will be changed in numpy array for initial array as well---")
arr = np.array([1,2,3,4,5])
sub_arr = arr[1:3]
print(sub_arr)
sub_arr[0] = 200

print(sub_arr)
print(arr)


print("---Value will be not be changed after copy function ---")
arr2 = np.array([1,2,3,4,5])
sub_arr2 = arr2[1:3].copy()
print(sub_arr2)
sub_arr2[0] = 200

print(sub_arr2)
print(arr2)
```
