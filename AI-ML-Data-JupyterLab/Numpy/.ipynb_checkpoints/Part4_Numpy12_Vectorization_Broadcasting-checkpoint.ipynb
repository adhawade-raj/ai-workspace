```python
import numpy as np
```

```python
# a has shape (2, 3).
# b has shape (3,).
# Broadcasting allows b to work across both rows of a.
# Vectorization allows NumPy to perform the addition without an explicit Python loop.
a = np.array([[1,2,3],[4,5,6]])
b = np.array([1,2,3])
print(a+b)

print("---Operation not allowed---")
print("----ValueError: operands could not be broadcast together with shapes (3,3) (2,3)----")
arr = np.array([[1,2,3],[1,2,3],[1,2,3]])
arr2 = np.array([[1,2,3],[1,2,3]])
print(arr+arr2)
```
