```python
import numpy as np
```

```python
arr = np.array([1,2,3,4,5])

print("---Math aggregate functions---")
print(np.sum(arr))
print(np.min(arr))
print(np.max(arr))


print("---Power functions---")
print(np.square(arr))


print("---Rounding functions---")
print(np.round(3.14))
print(np.ceil(3.14)) #to make round above

print("---Log functions---")
print(np.log(arr))
print(np.log10(arr))
```
