```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
np.random.seed(0)
scores = np.random.normal(70,10,100)

plt.hist(scores, color="skyblue", edgecolor="black", bins=10)

plt.axvline(x=55,
            label="Passing marks (55)",
            linestyle="--")

plt.ylabel("Frequency of students")
plt.xlabel("Scores")
```
