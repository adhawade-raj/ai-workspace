```python
import matplotlib.pyplot as plt
```

```python
# Pie Charts
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]
```

```python
plt.pie(amounts, labels=expenses)

# plt.style.available
```

```python
plt.pie(amounts, labels=expenses, autopct="%1.1f%%")
```

```python
# wedgeprops
plt.pie(amounts, labels=expenses, autopct="%1.1f%%",
       wedgeprops ={
           "edgecolor":"black",
           "linewidth":1,
           "linestyle":"--",
           # "fill":False
       }
       )
```

```python
# explode
explode = [0,0,0,0.2,0]

# wedgeprops
plt.pie(amounts, labels=expenses, autopct="%1.1f%%",
       wedgeprops ={
           "edgecolor":"black",
           "linewidth":1,
       },
        explode=explode,
        startangle=180, # startangle -> rotate pie by 90
        shadow=True #To give shadow on background
       )
```
