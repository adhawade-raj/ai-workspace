```python
import matplotlib.pyplot as plt
```

```python
movies = [
    "The Dark Knight ",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

plt.bar(years, oscar_revenue, color="black")
plt.title("Revenue Each Year")
plt.xlabel("Movies")
plt.ylabel("Revenue($M)")

# Use movies[0] (or 0) instead of years[0]
# plt.text(years[0], oscar_revenue[0] + 20, str(oscar_revenue[0]), ha="center")


for i in range(len(years)):
    plt.text(years[i], oscar_revenue[i]+20, str(oscar_revenue[i]), ha="center")


# to extend the height of y axis, so values will not tocuh border
plt.ylim(0, max(oscar_revenue) + 100)    
plt.show()
```
