```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = [7,8,5,6,9,4,10,15]

plt.boxplot(data)
plt.grid(True)
```

```python
# multiple dataset
group1 = np.random.normal(50,10,100)
group2 = np.random.normal(60,15,100)

# plt.boxplot([group1, group2], tick_labels=["group1", "group2"])
plt.boxplot([group1, group2], tick_labels=["group1", "group2"], whis=2.0, vert=False)
plt.grid(True)
plt.show()
```
