```python
import matplotlib.pyplot as plt
```

```python
# Scatter Plots
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

colors = ["red", "blue", "orange", "black", "orange", "Pink", "purple", "blue", "orange", "black"]

plt.scatter(age, bp, color=colors)
plt.title("AGE vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

plt.grid()
```

```python
# Logical Coloring

# Scatter Plots
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

colors = ["green" if x < 135 else "red" for x in bp]
sizes=bp

plt.scatter(age, bp, color=colors, s=bp)
# plt.scatter(age, bp, color=colors, s=bp, marker=">", alpha=0.5) #alpha = transparency
plt.title("AGE vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

plt.grid()
```

```python
# Plasma - color map (cmap)

# Logical Coloring

# Scatter Plots
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

colors = ["green" if x < 135 else "red" for x in bp]
sizes=bp

# plt.scatter(age, bp,s=bp, cmap="plasma", c=bp)
plt.scatter(age, bp,s=bp, cmap="OrRd", c=bp)
# plt.scatter(age, bp, color=colors, s=bp, marker=">", alpha=0.5) #alpha = transparency
plt.title("AGE vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.colorbar(label="BP")

# Annotation - annotate()
for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i], bp[i]))
    
plt.xlim(min(age), max(age)+10)
plt.ylim(min(bp), max(bp)+5)

```
