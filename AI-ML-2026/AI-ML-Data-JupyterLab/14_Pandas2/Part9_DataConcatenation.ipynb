```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    "id":[1,2,3],
    "name" : ["Adam", "Charlie", "Bob"]
})

df2 = pd.DataFrame({
    "id":[4,5,6],
    "name" : ["Glenn", "Max", "William"]
})
```

```python
# row wise concatenation
pd.concat([df1,df2])
```

```python
# clomn wise concatenation
pd.concat([df1,df2],axis=1)
```
