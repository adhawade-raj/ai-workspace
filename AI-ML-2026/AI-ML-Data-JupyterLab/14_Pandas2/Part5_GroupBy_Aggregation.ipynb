```python
import pandas as pd
import os

os.chdir(r"D:\Workspace\AI-Workspace\AI-ML-2026\AI-ML-Data-JupyterLab\14_Pandas2")
df = pd.read_csv("Part1_raw_data.csv")
df2 = df.copy()
```

```python
df2.groupby("country")["income"].mean()
```

```python
df2.groupby("country")["income"].min()
```

```python
df2.groupby("gender")["income"].max()
```

```python
# aggregate function
df2.groupby("gender")["income"].agg(["min", "max", "mean"])

# or
# df2.groupby("gender")["income"].aggregate(["min", "max", "mean"])
```

```python
# aggregate function witth aliases
df2.groupby("gender")["income"].agg(min_salary="min", max_salary="max", avg_salary="mean")
```

```python
# agg but diff functions on diff colomns
df2.groupby("country").agg({
        "income":"max",
        "age":"min"
})
```
