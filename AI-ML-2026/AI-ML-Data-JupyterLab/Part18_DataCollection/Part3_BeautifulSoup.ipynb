```python
from bs4 import BeautifulSoup
```

```python
with open(r"D:\Workspace\AI-Workspace\AI-ML-2026\AI-ML-Data-JupyterLab\Part18_DataCollection\data1.html", "r", encoding="utf-8") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml")    
```

```python
# soup
```

```python
# import sys
# !{sys.executable} -m pip install --user lxml
```

```python
all_h3 = soup.find_all("h3")
all_countries = []
for h3 in all_h3:
    name = (h3.getText(strip=True))

    # print(h3.find_parent("div")["class"])
    population = h3.find_next("div").select("span.country-population")[0].get_text(strip=True)

    all_countries.append([name,population])
```
