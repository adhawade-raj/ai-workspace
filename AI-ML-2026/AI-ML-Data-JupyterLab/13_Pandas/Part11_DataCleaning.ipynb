```python
# to fill age with mean values
df.fillna(0)
age_mean = df["age"].mean()
df["age"].fillna(age_mean)
```

```python
df.dropna() #raw drop
df.dropna(axis=1) #col drop
```

```python
# to store a data and create a backup/copy
cleaned_data = df.copy()
age_mean = cleaned_data["age"].mean()
cleaned_data["age"].fillna(age_mean)

cleaned_data
```
